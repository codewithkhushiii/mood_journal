```python
import os
import subprocess

MODEL = "Qwen/Qwen2.5-7B-Instruct-AWQ"

# Use a strong, private key. Configure the SAME key in your app.
API_KEY = "replace-with-a-long-random-secret"

env = os.environ.copy()
# FlashInfer's sampler JIT-compiles a CUDA kernel at startup and the link step fails on
# Kaggle ("/usr/bin/ld: cannot find -lcuda"). Use vLLM's built-in PyTorch sampler instead.
env["VLLM_USE_FLASHINFER_SAMPLER"] = "0"

log_file = open("/kaggle/working/vllm.log", "w")

server = subprocess.Popen(
    [
        "vllm", "serve", MODEL,
        "--host", "0.0.0.0",
        "--port", "8000",
        "--api-key", API_KEY,
        "--dtype", "half",
        "--max-model-len", "8192",
        "--gpu-memory-utilization", "0.90",
        "--max-num-seqs", "2",
    ],
    env=env,
    stdout=log_file,
    stderr=subprocess.STDOUT,
)

print("vLLM process started:", server.pid)
```

```python
import time
import requests
from kaggle_secrets import UserSecretsClient
from pyngrok import ngrok

secrets = UserSecretsClient()
ngrok_token = secrets.get_secret("NGROK_AUTHTOKEN")

ngrok.set_auth_token(ngrok_token)

# Wait until the local vLLM API is responding
for attempt in range(120):
    if server.poll() is not None:
        raise RuntimeError("vLLM exited. Check vllm.log.")

    try:
        r = requests.get(
            "http://127.0.0.1:8000/v1/models",
            headers={"Authorization": f"Bearer {API_KEY}"},
            timeout=2,
        )
        if r.ok:
            break
    except requests.RequestException:
        pass

    time.sleep(5)
else:
    raise TimeoutError("vLLM did not become ready.")

tunnel = ngrok.connect(8000, "http")
PUBLIC_URL = tunnel.public_url

print("API base URL:", PUBLIC_URL + "/v1")
```

```python
import requests, time
t = time.time()
r = requests.post(
    "http://127.0.0.1:8000/v1/chat/completions",
    headers={"Authorization": f"Bearer {API_KEY}"},
    json={
        "model": MODEL,
        "messages": [{"role": "user", "content": "Say hi in five words."}],
        "max_tokens": 32,
    },
    timeout=120,
)
print(r.status_code, round(time.time() - t, 1), "s")
print(r.json()["choices"][0]["message"]["content"])
```
